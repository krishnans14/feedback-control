## What do I mean?

I had a recent exchange with someone on a project whose domain appeared alien to my CV. And naturally the question of why am I interested in that project came up. I commented that, as a control engineer, I'm domain agnostic (engineering domain, that is). And a choice in a domain comes down to current personal preference, philosophy, what I consider as important, and such nebulous topics.

This domain agnosticism can be misunderstood easily to be the same as looking at everything through a black-box perspective. That is, get input and output data from the system, and construct a model using this data by subjecting it through standard practices and that's it. Most people would agree that this does not work. And it leads to pretty dubious effects in complex systems such as the society. This is why, for instance, aspects such as ethics, bias, etc., are given strong emphasis by those who attempt to disseminate AI skills to a wide range of audience. For example, Fast.ai has a dedicated [lecture on Ethics](https://course.fast.ai/videos/?lesson=5).

But as a control engineer, my interest is in engineering problems. While the complexity of some engineering problems have been increasing with the incorporation of humans in the loop (Automated cars), a good part of the systems have reliable models. To represent engineering systems, humans use mathematical tools, just like we use them to represent nature.

And that my interest is on the problems that are navigable 

Recently, I was exchanging with someone about why I want to work on a particular project. I commented that as a control engineer, I can work in any domain and the relevance is on the tools I possess. So choosing a project comes down to personal preference. I then went onto explain why this project is within the space of my current personal preference and interest. 

Control engineers see everything as a system with inputs, outputs and internal dynamics/complexities. A black-box, if you will. And a large class of technological problems fall under this umbrella. The success of deep neural network in a lot of application owes to this black-box perspective to solve complex problems such as image classification. 

This is not to discount the importance of immersion into domains in pushing technology forward. There are several types of problems that simply can not be tackled solely using the black-box approach. But control engineers focus on problems for which domain skills are required, but not with decades of immersion. Nevertheless, any problem requires a *good* understanding of the domain. This is why, working with multiple domains, control engineers pick up the knack of immersion into domains through a top-down approach. They are trained (not necessarily by their diplomas and degrees, but by experience and practice) to immerse into those domain aspects that are useful to successfully solve problems of interest.

:::{.callout-tip}

If you think you have this knack, you have a control engineer inside you.

:::

Beyond engineering systems, this problem-solving perspective is deployed by the more generic *system thinkers* who tackle much more complex problems that pervade the human-centric worldview.

## Prelude

My preference to domain hence comes solely on my personal choices which vary with time and mood. It is likely driven by the current personal philosophy and what I think matters. Electronics enthralled me during my undergraduate studies (even though my discipline was Instrumentation and Control with exposure to almost all domains) and I was determined to work in this domain. And I spent the next 7 years associated with Electronics (4 years at Wipro Technologies and a 3-years-stint as a lab assistant in an electronics lab at IIT Bombay as I bid adieu to Electronics).

Ever since my masters, energy systems have been of interest to me with my contributions [household Electric Load Disaggregation](https://krishnans14.github.io/feedback-control/projects/2020/12/30/NILM.html) while in TCS and my thesis that ventured into optimizing building energy.

The exposure to water networks during my [postdoctoral research project](https://krishnans14.github.io/feedback-control/projects/2021/01/08/Project-SPHEREAU.html) made me sit up to notice that my interest may go deeper that energy domain because energy drives almost all domains.

This kicked off a bit of my recent reflections on what threads connect my past projects. I saw the following:

* Problem solving in existing systems (as against designing something new from scratch because more time is spent in following standards)
* Large and complex systems (and hence emergent properties and problems to work on)
* Assimilating knowledge spread over multiple individuals/teams (can't work alone in my garage to invent something)
* Don't break what works (fancy ideas would get booted out with impunity because it may break things)

My interest in this article is to show how the above notions have run as a silent thread through the different engineering problems I have worked on.
